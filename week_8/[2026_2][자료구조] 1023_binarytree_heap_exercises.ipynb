## Part 3. 힙 구현
- Max Heap: 완전 이진트리 + 부모 ≥ 자식 → 루트 `h[0]`가 최댓값
- 형제끼리 순서는 없다 (루트만 보장)
```
         93            h = [93, 38, 27, 29, 31, 15]
       /    \
     38      27
    /  \    /
   29  31  15
```
`is_max_heap`은 확인용 (완성)

````python
def is_max_heap(h):
    # 모든 i ≥ 1에 대해 부모 ≥ 자식인지 확인
    for i in range(1, len(h)):
        if h[(i - 1) // 2] < h[i]:
            return False
    return True


print(is_max_heap([93, 38, 27, 29, 31, 15]))   # True
print(is_max_heap([93, 38, 27, 29, 31, 40]))   # QUIZ: 15를 40으로 바꾸면? → False (부모 27보다 커짐)
````

### Q6. MaxHeap - 삽입 `push(x)`
1. `append`로 마지막 자리에 추가
2. `i = len(h) - 1`, `i > 0` 동안: `p = (i - 1) // 2` → `h[i] <= h[p]`면 멈춤, 아니면 교환 후 `i = p`

`peek()`: 비어있으면 None, 아니면 `h[0]`

````python
class MaxHeap:
    def __init__(self):
        self.h = []              # list 하나로 Heap 저장

    def size(self):
        return len(self.h)

    def is_empty(self):
        return len(self.h) == 0

    def peek(self):
        # TODO
        if not self.h:
            return None
        return self.h[0]

    def push(self, x):
        # TODO: 위 순서대로 구현하세요.
        self.h.append(x)
        i = len(self.h)-1
        while i > 0:
            p = (i - 1) // 2
            if self.h[i] <= self.h[p]:
                break
            self.h[i], self.h[p] = self.h[p], self.h[i]
            i = p
````

### Q13. 지금 자료구조가 효율적일까?

#### (1) 프린터 대기열
> "먼저 보낸 문서부터 출력되게 해줘."

````python
import time

class Printer:
    def __init__(self):
        self.jobs = []

    def add_job(self, name):
        self.jobs.append(name)

    def print_next(self):
        if self.jobs:
            return self.jobs.pop(0)


pr = Printer()
for name in ["과제.pdf", "보고서.docx", "사진.jpg"]:
    pr.add_job(name)
print(pr.print_next(), pr.print_next(), pr.print_next())   # 보낸 순서대로 나온다

# 문서가 2배가 되면 시간은?
for n in [20_000, 40_000]:
    pr = Printer()
    for i in range(n):
        pr.add_job(f"문서{i}")
    start = time.perf_counter()
    while pr.jobs:
        pr.print_next()
    print(f"문서 {n}개 출력: {time.perf_counter() - start:.3f}초")
````

### Q14. 버그가 있는 힙 코드 고치기
`BuggyMaxHeap`에 틀린 곳이 2군데 있다. 실행해서 증상을 확인하고 코드를 직접 고치세요.

````python
class BuggyMaxHeap:
    def __init__(self):
        self.h = []

    def push(self, x):
        self.h.append(x)
        i = len(self.h) - 1
        while i > 0:
            p = (i - 1) // 2
            if self.h[i] <= self.h[p]:
                break
            self.h[i], self.h[p] = self.h[p], self.h[i]

    def pop(self):
        if not self.h:
            return None
        root = self.h[0]
        last = self.h.pop()
        if self.h:
            self.h[0] = last
            i, n = 0, len(self.h)
            while True:
                l = 2 * i + 1
                if l < n and self.h[l] > self.h[i]:
                    self.h[i], self.h[l] = self.h[l], self.h[i]
                    i = l
                else:
                    break
        return root
````
