### Q1. [배열] 맨 앞에서 꺼내는 과정을 직접 구현해보기
`list.pop(0)` 이 내부적으로 하는 일을 반복문으로 직접 만들어보기

순서: ① 맨 앞 값을 따로 저장한다 → ② 1번 칸부터 끝까지, 각 값을 한 칸씩 **앞으로** 옮긴다 → ③ 맨 뒤 칸을 제거한다 → ④ 저장해둔 값을 반환한다.

````python
def dequeue_front_array(arr):
    if len(arr) == 0:
        return None
    value = arr[0]             # ① 맨 앞 값 저장
    n = len(arr)
    # TODO: range(1, n) 으로 1번 칸부터 끝까지 가면서
    for i in range(1, n):
        arr[i-1] = arr[i]
    #       arr[i-1] = arr[i] 로 한 칸씩 앞으로 옮기세요.

    arr.pop()                  # ③ 맨 뒤 칸(이제 중복된 값) 제거
    return value               # ④ 저장해둔 값 반환
````

### Q3. 연결리스트로 큐 만들기 (LinkedQueue)
지난 실습의 `Node`를 다시 만들어서, list 없이 연결리스트로 큐 만들기

힌트: 스택은 `head` 하나로 충분했지만, 큐는 **앞(front)에서 꺼내고 뒤(rear)에 넣기** 때문에 양쪽 끝을 모두 기억
- `enqueue` → `rear` 뒤에 새 노드를 붙이고 `rear`를 옮김 (O(1))
- `dequeue` → `front`의 값을 꺼내고 `front`를 다음 노드로 옮김

````python
class Node:
    def __init__(self, value):
        # TODO: self.value에 value를 저장하고, self.next를 None으로 초기화하세요.
        self.value = value
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None   # 맨 앞 노드 (꺼내는 쪽)
        self.rear = None    # 맨 뒤 노드 (넣는 쪽)
        self.count = 0

    def enqueue(self, value):
        new_node = Node(value)
        if self.rear is None:
            # TODO: 큐가 비어있으면 front와 rear 모두 new_node로 만드세요.
            self.front = new_node
            self.rear = new_node
        else:
            # TODO: 기존 rear.next가 new_node를 가리키게 하고, rear를 new_node로 옮기세요.
            self.rear.next = new_node
            self.rear = new_node
        self.count += 1

    def dequeue(self):
        if self.front is None:
            return None
        # TODO: front의 value를 저장하고, front를 front.next로 옮기세요.
        #       옮긴 뒤 front가 None이 되었다면(큐가 완전히 비었다면) rear도 None으로 만드세요.
        #       count를 1 줄이고, 저장해둔 값을 반환하세요.
        
        val = self.front.value
        self.front = self.front.next
        if self.front is None:
            self.rear = None   
        self.count -= 1
        return val

    def peek(self):
        # TODO: front가 None이면 None, 아니면 front.value를 반환하세요.
        if self.front == None:
            return None
        else:
            return self.front.value

    def is_empty(self):
        # TODO: front가 None이면 True, 아니면 False를 반환하세요.
        return self.front is None

    def size(self):
        return self.count

    def print_all(self):
        current = self.front
        # current를 따라가며 값을 "front -> 10 -> 20 -> rear" 형태로 한 줄 출력
        print("front", end=" -> ")
        while current is not None:
            print(current.value, end=" -> ")
            current = current.next
        print("rear")
````

### Q4. 스택 두 개로 큐 만들기 (StackQueue)
스택(LIFO)을 두 번 뒤집으면 큐(FIFO)가 된다는 고전 문제.

- `in_stack`: enqueue할 때 그냥 push.
- `out_stack`: dequeue할 때 여기서 pop. 비어있으면 `in_stack`의 것을 전부 pop해서 `out_stack`에 push (순서가 뒤집힘 → 맨 처음 넣은 게 맨 위로 온다).

````python
class StackQueue:
    def __init__(self):
        self.in_stack = []    # list를 스택으로 사용 (append / pop)
        self.out_stack = []

    def enqueue(self, value):
        self.in_stack.append(value)

    def dequeue(self):
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())
        if not self.out_stack:
            return None
        return self.out_stack.pop()
````

## Part 2. 원형 큐(Circular Queue) - 배열로 O(1) 큐 만들기

배열로 큐를 만들 때 `pop(0)` 대신 **`front` 인덱스만 한 칸 앞으로 옮기면** 값을 당길 필요가 없어서 O(1)이 된다.

```
data  = [A, B, C, _, _]      front=0, rear=3
dequeue → data = [_, B, C, _, _]   front=1  (A 자리는 비워두고 front만 이동)
```

하지만 이렇게 하면 앞쪽에 **빈 칸이 쌓여서 결국 뒤쪽 공간이 모자라게 된다.**

해결책: `rear`가 배열 끝에 닿으면 **다시 0번 칸으로 돌아가서** 앞쪽의 빈 칸을 재활용한다. → 배열을 **원형(ring)** 처럼 쓰는 것.

핵심 공식: `다음 인덱스 = (현재 인덱스 + 1) % capacity`

```
capacity=5          idx: 0  1  2  3  4
enqueue A,B,C,D,E  →    [A, B, C, D, E]   front=0, rear=0(한 바퀴 돌아서), count=5 (가득 참)
dequeue, dequeue   →    [_, _, C, D, E]   front=2
enqueue F          →    [F, _, C, D, E]   rear=1  ← 0번 칸을 재활용!
```

- `front`: 다음에 꺼낼 칸의 인덱스
- `rear` : 다음에 넣을 칸의 인덱스
- `count`: 현재 들어있는 개수 (이게 있으면 '비었음'과 '가득 참'을 쉽게 구분할 수 있다)

````python
class CircularQueue:
    def __init__(self, n):
        self.data = [None] * n     # 크기는 배열이 들고 있음
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def is_full(self):
        return self.count == len(self.data)

    def enqueue(self, value):
        if self.is_full():
            return False
        self.data[self.rear] = value
        self.rear = (self.rear + 1) % len(self.data)
        self.count += 1
        return True

    def dequeue(self):
        if self.is_empty():
            return None
        value = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % len(self.data)
        self.count -= 1
        return value

    def peek(self):
        return None if self.is_empty() else self.data[self.front]

    def print_all(self):
        print("front", end=" -> ")
        for i in range(self.count):
            print(self.data[(self.front + i) % len(self.data)], end=" -> ")
        print("rear")

    def show(self):
        print(f"data={self.data}  front={self.front}  rear={self.rear}  count={self.count}")
````

### Capacity 없이 원형 큐 만들기 (연결리스트)

````python
class CircularQueue_ver2:
    def __init__(self):
        self.rear = None      # 마지막 노드. rear.next 가 곧 맨 앞(front)
        self.count = 0

    def is_empty(self):
        return self.rear is None

    def enqueue(self, value):
        new_node = Node(value)
        if self.rear is None:
            new_node.next = new_node        # 혼자서 원을 이룸 (자기 자신을 가리킴)
        else:
            new_node.next = self.rear.next  # 새 노드 → front
            self.rear.next = new_node       # 기존 rear → 새 노드
        self.rear = new_node
        self.count += 1

    def dequeue(self):
        if self.rear is None:
            return None
        front = self.rear.next
        value = front.value
        if front is self.rear:              # 노드가 하나뿐이면 비게 됨
            self.rear = None
        else:
            self.rear.next = front.next     # front를 건너뛰어 원에서 뺌
        self.count -= 1
        return value

    def peek(self):
        return None if self.rear is None else self.rear.next.value

    def size(self):
        return self.count

    def print_all(self):
        print("front", end=" -> ")
        if self.rear is not None:
            current = self.rear.next
            for _ in range(self.count):
                print(current.value, end=" -> ")
                current = current.next
        print("rear")
````

## Part 4. collections.deque
파이썬은 `collections.deque`를 기본 제공하고, 이게 곧 **큐 + 덱** 역할을 한다.
- `append` / `popleft` → 큐
- `append` / `pop` → 스택
- `appendleft` / `pop` 등 → 덱
- `maxlen`을 주면 **고정 크기 원형 큐**처럼 동작 (가득 차면 반대쪽에서 자동으로 밀려남)
- `rotate(k)` → 원형으로 k칸 회전

````python
from collections import deque

dq = deque()
dq.append("A"); dq.append("B"); dq.append("C")
print(dq.popleft())    # A  (큐처럼)
print(dq.pop())        # C  (스택처럼)
dq.appendleft("Z")
print(dq)              # deque(['Z', 'B'])

# maxlen: 최근 3개만 기억하는 원형 큐 (ex. 최근 본 상품 3개)
recent = deque(maxlen=3)
for item in ["신발", "모자", "가방", "시계"]:
    recent.append(item)
print(recent)          # deque(['모자', '가방', '시계'], maxlen=3)  ← '신발'은 밀려남

# rotate: 원형으로 돌리기
ring = deque([1, 2, 3, 4, 5])
ring.rotate(-2)        # 왼쪽으로 2칸 (앞의 두 개가 뒤로 감)
print(ring)            # deque([3, 4, 5, 1, 2])
````
