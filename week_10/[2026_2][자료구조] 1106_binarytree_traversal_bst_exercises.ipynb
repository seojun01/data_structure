`left`, `right`로 자식을 참조한다. 연결하지 않은 자식은 None.

````python
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def show(node, level=0):
    # 트리를 옆으로 눕혀서 출력 (고개를 왼쪽으로 돌려서 보기)
    if node is None:
        return
    show(node.right, level + 1)
    print("      " * level + str(node.data))
    show(node.left, level + 1)
````

실행해서 매 반복마다 꺼낸 노드와 큐에 남은 값을 확인

````python
def level_order_trace(root):
    queue = deque([root])
    while queue:
        node = queue.popleft()
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)
        print(f"꺼냄 {node.data:>3}  | 큐: {[n.data for n in queue]}")


level_order_trace(T)
````

### Q11. Queue 대신 Stack을 쓰면?
`popleft()` → `pop()`. 실행 전에 순서를 먼저 예상해 보세요.

위의 큐 상태 보기와 같은 형식으로 스택 상태를 출력한다. (스택은 **오른쪽 끝이 top**)

````python
def stack_order(root):
    stack = [root]
    result = []
    while stack:
        node = stack.pop()          # 나중에 넣은 것부터
        result.append(node.data)
        if node.left:
            stack.append(node.left)
        if node.right:
            stack.append(node.right)
        print(f"꺼냄 {node.data:>3}  | 스택: {[n.data for n in stack]}")
    return result


print(stack_order(T))
````

### Q25. 메뉴 프로그램으로 실행해 보기
완성한 `BST` 클래스를 메뉴로 사용한다. (`menu` 함수는 제공)

````python
def menu(bst, inputs=None):
    # inputs : 입력을 미리 넣어 자동 실행할 때 쓰는 리스트 (없으면 직접 입력)
    it = iter(inputs) if inputs is not None else None

    def read(prompt):
        if it is None:
            return input(prompt)
        v = next(it)
        print(prompt + v)
        return v

    while True:
        print()
        print("*** Binary Search Tree(BST) ***")
        print("1. 노드 삽입")
        print("2. 노드 삭제")
        print("3. 노드 검색")
        print("4. 최댓값 구하기")
        print("5. 최솟값 구하기")
        print("6. 이진 검색 트리 출력(중위순회)")
        print("0. 프로그램 종료")
        choice = read("메뉴 선택 : ")

        if choice == "1":
            bst.insert(int(read("삽입할 값 : ")))
        elif choice == "2":
            bst.delete(int(read("삭제할 값 : ")))
        elif choice == "3":
            found, path = bst.search(int(read("찾을 값 : ")))
            print("찾았다!" if found else "없다.", "경로:", path)
        elif choice == "4":
            print("최댓값 :", bst.find_max())
        elif choice == "5":
            print("최솟값 :", bst.find_min())
        elif choice == "6":
            print("중위 순회 :", bst.inorder())
            show(bst.root)
        elif choice == "0":
            print("프로그램 종료")
            break
        else:
            print("잘못된 메뉴입니다.")
````

````python
menu(BST())
````
