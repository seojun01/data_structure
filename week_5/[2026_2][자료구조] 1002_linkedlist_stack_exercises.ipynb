### Q1. [배열] 맨 앞에 삽입하는 과정을 직접 구현해보기
`list.insert(0, value)` 가 내부적으로 하는 일을 반복문으로 직접 만들어보기

순서: ① 리스트 맨 뒤에 빈 자리를 하나 늘린다 → ② 맨 뒤부터 거꾸로 한 칸씩 옆 칸의 값을 옮긴다 → ③ 맨 앞 칸에 value를 넣는다.

```python
def insert_front_array(arr, value):
    arr.append(None)          # ① 칸 하나 확보 (아무 값이나 넣어서 크기만 늘림)
    n = len(arr)
    # TODO: range(n-1, 0, -1) 로 맨 뒤부터 거꾸로 내려오면서
    for i in range(n-1, 0, -1):
    #       arr[i] = arr[i-1] 로 한 칸씩 뒤로 옮기세요.
        arr[i] = arr[i-1]
    arr[0] = value             # ③ 맨 앞 칸에 value 채우기
    return arr
```

### Step 4. Q3. append(value) - 맨 뒤에 노드 추가
리스트가 비어있으면 새 노드를 head로 만들고, 

비어있지 않으면 맨 마지막 노드까지 따라가서 그 뒤에 새 노드를 연결하기.

```python
def append(self, value):
    new_node = Node(value)
    if self.head is None:
        # 리스트가 비어있으면 head를 new_node로 만든다.
        self.head = new_node
        return

    # 마지막 노드까지 따라간 뒤 그 뒤에 new_node를 연결한다.
    current = self.head
    while current.next is not None:
        current = current.next
    current.next = new_node


LinkedList.append = append # LinkedList 클래스의 메서드로 등록
```

### Step 6. Q5. insert_front(value) - 맨 앞에 노드 추가 (O(1))

### Step 7. Q6. search(value) - 특정 값 찾기

```python
def search(self, value):
    current = self.head
    # current를 따라가며 값이 같은지 확인한다.
    while current is not None:
        if current.value == value:
            return True
        current = current.next

    return False


LinkedList.search = search
```

### Step 8. Q7. delete_front() - 맨 앞 노드 삭제하고 값 반환

```python
def delete_front(self):
    if self.head is None:
        return None
    # head의 value를 저장하고, head를 다음 노드로 옮긴 뒤 저장해둔 값을 반환한다.
    value = self.head.value
    self.head = self.head.next
    return value


LinkedList.delete_front = delete_front
```

## Part 4. 실전에서는 `deque`를 씁니다
지금까지 만든 것들은 원리를 이해하기 위한 연습.
실제로 파이썬은 `collections.deque`를 기본 제공함.
`deque`는 내부적으로 **양방향 연결리스트**로 구현되어 있어서, 양쪽 끝에서 추가/삭제가 모두 O(1). 위에서 만든 것과 원리가 같음.

```python
from collections import deque

dq = deque([1, 2, 3])
dq.append(4)         # 맨 뒤에 추가 (O(1))
dq.appendleft(0)      # 맨 앞에 추가 (O(1))
print(dq)             # deque([0, 1, 2, 3, 4])

print(dq.pop())        # 맨 뒤에서 꺼내기 (O(1)) -> 4
print(dq.popleft())     # 맨 앞에서 꺼내기 (O(1)) -> 0
print(dq)
```

```python
class MusicPlayer:
    def __init__(self):
        self.list = DoublyLinkedList()
        self.current = None

    def add_song(self, name):
        self.list.append(name)
        if self.current is None:
            self.current = self.list.head   # 첫 곡은 자동으로 현재 곡이 됨

    def current_song(self):
        return self.current.value if self.current else None

    def next_song(self):
        # TODO: self.current.next가 있으면(None이 아니면) self.current를 그쪽으로 이동하세요.

        return self.current_song()

    def prev_song(self):
        # TODO: self.current.prev가 있으면(None이 아니면) self.current를 그쪽으로 이동하세요.

        return self.current_song()
```
